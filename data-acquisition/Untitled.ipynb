```python
import pandas as pd

url = "owid-energy-data.csv"

# 1. Inspect headers first
cols = pd.read_csv(url, nrows=0).columns.tolist()

# 2. Define required columns
wanted = [
    "country", "year", "iso_code", "population",
    "primary_energy_consumption", "electricity_demand",
    "renewables_consumption", "fossil_fuel_consumption"
]

# 3. Load only specified columns
wanted_cols = [c for c in wanted if c in cols]
energy = pd.read_csv(url, usecols=wanted_cols)

# Inspection & Memory usage
full_energy_shape = pd.read_csv(url, nrows=1).shape[1] # 130 columns
memory_mb = energy.memory_usage(deep=True).sum() / 1024**2
print(f"Loaded Shape: {energy.shape} | Memory Usage: {memory_mb:.2f} MB")
```
